```python

from transformers import pipeline

# Contexte initial
context = (
    "The Mona Lisa is a famous painting by Leonardo da Vinci. "
    "It is displayed at the Louvre Museum in Paris."
)

# Question initiale
question = "Who painted the Mona Lisa?"

qa = pipeline(
    task="question-answering",
    model="distilbert/distilbert-base-cased-distilled-squad"
)

# Prédiction
output = qa(question=question, context=context)

print(output["answer"])
```

```python
from transformers import pipeline

context = (
    "The Mona Lisa is a famous painting by Leonardo da Vinci. "
    "It is displayed at the Louvre Museum in Paris."
)

question = "Who painted the Mona Lisa?"

# Charger GPT-2
gpt2 = pipeline(
    task="text-generation",
    model="openai-community/gpt2"
)

prompt = f"{context}\nQuestion: {question}\nAnswer:"

output = gpt2(
    prompt,
    max_new_tokens=30,
    do_sample=False,
    pad_token_id=gpt2.tokenizer.eos_token_id
)

print(output[0]["generated_text"])
```

```python

from transformers import AutoConfig

modeles = {
    "BERT": "bert-base-uncased",
    "GPT-2": "gpt2",
    "T5": "t5-small"
}

for nom, identifiant in modeles.items():
    config = AutoConfig.from_pretrained(identifiant)

    print("Modèle :", nom)
    print("Type :", config.model_type)
    print("Classes :", config.architectures)
    print("Encoder-decoder :", config.is_encoder_decoder)
    print("Is decoder :", config.is_decoder)

    print("Couches :", getattr(
        config, "num_hidden_layers",
        getattr(config, "n_layer", None)
    ))

    print("Dimension :", getattr(
        config, "hidden_size",
        getattr(config, "n_embd",
                getattr(config, "d_model", None))
    ))

    print("-" * 40)
```

```python

from transformers import pipeline

texte = "Artificial intelligence helps smart buildings reduce electricity consumption."

# GPT-2 : génération de texte
gpt2 = pipeline("text-generation", model="gpt2")

# T5 : transformation de texte
t5 = pipeline("text2text-generation", model="t5-small")

# Résultat GPT-2
resultat_gpt2 = gpt2(
    texte,
    max_new_tokens=35,
    do_sample=False,
    pad_token_id=gpt2.tokenizer.eos_token_id
)

# Résultat T5 : traduction
resultat_t5 = t5(
    "translate English to German: " + texte,
    max_new_tokens=50,
    do_sample=False
)

print("GPT-2 :", resultat_gpt2[0]["generated_text"])
print("T5 :", resultat_t5[0]["generated_text"])
```
